```python
#Program that prints reverse of a string
def reverse(s):
  return s[::-1]
print(reverse("india"))
```

```python
#without using indexing and function
def reverse(str):
  for i in range(len(str)):
    print(str[len(str)-i-1],end="")
reverse("India")
```
