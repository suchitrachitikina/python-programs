```python
#List Operations
L1=[1,"mani",2,3]
L2=[1,8,8,8]
print(L1+L2)
print(L1*2)
print(L2[2])
print(L1[1:4])
if 8 in L2:
  print("YES")
for i in L1:
  print(i)
for i in range(len(L2)):
  print(L2[i])
```

```python
#List Functions
L1=[3,5,7,1,9]
print(list(reversed(L1)))
print(sorted(L1))
```

```python
#List Methods
L1=[2,8,9,6,5]
L2=[34,89,70,65,89]
L1.append("Hello")
print(L1)
L1.extend(L2)
print(L1)
L2.insert(0,"Hi")
print(L2)
L2.pop(2)
print(L2)
L2.remove(34)
print(L2)
d=L1.index(9)
print(d)
c=L1.count(89)
print(c)
L3=[4,9,0,1,6]
L3.sort()
print(L3)
```

```python
#Sum of elements in a list
L1=[1,2,3,4,5]
sum=0
for i in L1:
  sum=sum+i
print(sum)
```

```python
# print elements in odd indexs
L1=[54,67,89,45,83,92,10,50,60]
for i in range(0,len(L1),2):
  print(L1[i])


```

```python
#print the even elements
L1=[2,3,4,5,6,7,89]
for i in L1:
  if(i%2==0):
    print(i)
```

```python
#List comprehension
n=int(input())
L=[i**2 for i in range (n)]
print(L)
```

```python
L1=[i**2 for i in L]
print(L1)
L2=[int(i) for i in input().split()]
print(L2)
```
