```python
#if-elif-else statements
marks = int(input("Enter marks: "))

if marks >= 90 and marks <= 100:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
elif marks >= 0:
    print("Fail")
else:
    print("Invalid marks")
```

```python
#if-elif-else statements
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print("Electricity Bill:", bill)
```

```python
#Nested if statement
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))
if(a==b):
  if(a==c):
    print("EQUILATERAL TRIANGLE")
  else:
    print("ISOSCELES TRIANGLE")
else:
  if(a!=c):
    print("SCALANE TRIANGLE")
  else:
    print("ISOSCALES TRIANGLE")
```

```python
#MATCH CASE
day=int(input("Enter any Number(1 to 7):"))
match day:
  case 1:
    print("Sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("Thursday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
  case _:
    print("Invalid")
```

```python
#PROGRAM THAT SWAPS TWO VARIABLES WITHOUT USING TEMPORARY VARIABLE
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
print("BEFORE SWAPPING:")
print("a = ",a)
print("b = ",b)
a,b=b,a
print("AFTER SWAPPING:")
print("a = ",a)
print("b = ",b)
```
