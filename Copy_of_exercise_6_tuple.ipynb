```python
#Tuple Operations
t=(1,2,3,4,"Hello")
t1=(1,9,0,8)
print(t+t1)
print(t1*5)
for i in t1:
  print(i)
if 8 in t1:
  print("Yes")
```

```python
#Tuple Methods
t1=(1,2,3,1,5,6,7,5)
print("count of 1:",t1.count(1))
print("index of 7:",t1.index(7))
```

```python
#Tuple Functions
t=(1,9,6,7,8,9)
t1=(3,4,5,6)
print(len(t))
print(min(t))
print(max(t))
print(sum(t))
print(sorted(t))
print(tuple(sorted(t)))
print(tuple(zip(t,t1)))
```

```python
# write a program to count vowels in a string(no control flow allow)
s=str(input())
s1=s.lower()
vowel_count=s1.count('a')+s1.count('e')+s1.count('i')+s1.count('o')+s1.count('u')
print( "Number of Vowels in a given string:",vowel_count)
```

```python
#Set operations
a={1,2,3}
b={3,4,5}
print(a|b)
print(a&b)
print(a-b)
print(a^b)
```

```python
#set methods
s={1,2,3,5}
s.add(9)
print(s)
s.update((8,7,4,6))
print(s)
s.remove(3)
print(s)
s.discard(5)
print(s)
s.pop()
print(s)
s.clear()
print(s)
print(s.union({1,2,0}))
print(s.intersection((1,2,9)))
print(s.difference((1,2,9)))
print(s.symmetric_difference((1,2,9)))
```
