```python
d={"name": "Suchi","sno":20,"grade":9.8}
print(d)
```

```python
print(d["name"])
d["Branch"]="AI&DS"#Adding another value
print(d)
d["sno"]=18 #modifying the value
print(d)
```

```python
#Built in functions
print(len(d))
print(sorted(d))
print(max(d))
print(list(d))
```

```python
# dictonary Methods
d.get("name")
d.keys()
print(d)
d.values()
d.items()
d.setdefault("age",18)
d.update({"Year":2,"age":18})
print(d)
```
